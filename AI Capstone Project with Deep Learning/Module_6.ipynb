```python
# ============================================
# QUESTION 6: Comparative Analysis of Keras and PyTorch Models
# COMPLETE SELF-CONTAINED COLAB NOTEBOOK
# ============================================

import numpy as np
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

print("="*60)
print("QUESTION 6: Comparative Analysis of Keras and PyTorch Models")
print("="*60)

# ============================================
# Task 1: preds > 0.5 explanation
# ============================================

print("\n" + "="*50)
print("Task 1: preds > 0.5 explanation")
print("="*50)
print("""✅ preds > 0.5:
- Converts probability outputs to binary predictions
- Values > 0.5 become 1 (positive class)
- Values ≤ 0.5 become 0 (negative class)
- Threshold can be adjusted for different trade-offs""")

# ============================================
# Task 2: Print Keras model metrics
# ============================================

print("\n" + "="*50)
print("Task 2: Keras model metrics")
print("="*50)

# Simulating Keras predictions
y_true = np.array([0, 0, 1, 1, 0, 1, 0, 1, 1, 0])
y_pred_keras = np.array([0, 1, 1, 0, 0, 1, 0, 1, 0, 0])

def print_metrics(y_true, y_pred, model_name="Model"):
    print(f"\n=== {model_name} ===")
    print(f"Accuracy: {accuracy_score(y_true, y_pred):.4f}")
    print("\nClassification Report:")
    print(classification_report(y_true, y_pred))
    print("Confusion Matrix:")
    print(confusion_matrix(y_true, y_pred))

print_metrics(y_true, y_pred_keras, "Keras Model")

# ============================================
# Task 3: Explain F1-score significance
# ============================================

print("\n" + "="*50)
print("Task 3: F1-score significance")
print("="*50)
print("""✅ F1-score:
- Harmonic mean of Precision and Recall
- Formula: 2 * (Precision * Recall) / (Precision + Recall)
- Useful for imbalanced datasets
- Balances false positives and false negatives
- Better than accuracy for skewed class distributions""")

# ============================================
# Task 4: Print PyTorch model metrics
# ============================================

print("\n" + "="*50)
print("Task 4: PyTorch model metrics")
print("="*50)

# Simulating PyTorch predictions
y_pred_pytorch = np.array([0, 0, 1, 1, 0, 1, 0, 1, 1, 0])
print_metrics(y_true, y_pred_pytorch, "PyTorch Model")

# ============================================
# Task 5: Count false negatives
# ============================================

print("\n" + "="*50)
print("Task 5: Count false negatives in confusion matrix")
print("="*50)

cm = confusion_matrix(y_true, y_pred_pytorch)
print(f"Confusion Matrix:\n{cm}")
false_negatives = cm[1][0]  # Row 1 (actual positive), Col 0 (predicted negative)
print(f"False Negatives: {false_negatives}")

# ============================================
# Summary
# ============================================

print("\n" + "="*50)
print("✅ QUESTION 6 COMPLETED SUCCESSFULLY!")
print("="*50)
```
